```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("bike_sales_india.csv")
brand_colors = {
    'Kawasaki': '#32CD32',    # Lime Green
    'Yamaha': '#00008B',      # Racing Blue
    'KTM': '#FF6600',         # Ready-to-Race Orange
    'Royal Enfield': '#333333', # Gunmetal Grey
    'Hero': '#D32F2F',        # Red
    'TVS': '#1976D2',         # Blue/White
    'Honda': '#B71C1C',       # Red
    'Bajaj': '#1A237E'        # Indigo
}
```

```python

# Analysis of Brand and Model Data
B_d = df["Brand"].value_counts().reset_index(name="Total")
plt.figure(figsize=(25,9))
plt.subplot(1,2,1)
sns.barplot(data=B_d,y="Brand",x="Total",hue ="Brand", palette=brand_colors)
plt.ylabel("Brand Name")
plt.title("Top Bike Brands Based on Data")
plt.subplot(1,2,2)
M_d = df["Model"].value_counts().reset_index(name="Total").head(10)
sns.barplot(M_d,x="Total",y="Model",hue="Model" , palette="viridis")
plt.ylabel("Model Name")
plt.title("Top 10 popular Models")
plt.suptitle("Brand and Model Analysis")
plt.show()
```

### Insights
- Brand distribution is relatively uniform  
- No single brand dominates  
- Kawasaki leads slightly but difference is small  

- It is same with the Top 10 models Models too
- Ray ZR leads slightly


```python
#Geographic Analysis 
g_a = df["State"].value_counts().reset_index(name = "Total")
plt.figure(figsize=(25,10))

plt.subplot(1,2,1)
sns.barplot(data=g_a, y="State", x="Total",hue="State" , palette="gist_rainbow")
plt.title("Total Records by State")

plt.subplot(1,2,2)
sns.countplot(data=df, y="State", hue="City Tier")
plt.title("State vs City Tier Distribution")

plt.show()

```

## Insights

-Bike records are fairly evenly distributed across major states, with no extreme concentration in a single region.
-Punjab has a slightly more distribution

- well the distribution of bikes around different city tiers changes accroding to the state.
- Tier 3 have slightly more distributions compared to the other City Tier

```python
# Fuel Analysis
f_a = df["Fuel Type"].value_counts().reset_index(name = "Total")
plt.title("Distribution by Fuel Type")
sns.barplot(f_a,x="Total",y="Fuel Type",hue= "Fuel Type",palette="magma")
```

## Insights
- Petrol and Hybrid bikes dominate the dataset, with Electric bikes slightly lower but still significant.

```python
# Price and Resale trends
plt.figure(figsize=(20,9))
plt.subplot(1,2,2)
sns.scatterplot(x="Price (INR)", y="Resale Price (INR)", data=df, alpha=0.5)
plt.title("Price vs Resale trend")
plt.subplot(1,2,1)
sns.histplot(df["Price (INR)"], bins=30, kde=True)
plt.title("Price Distribution")
plt.show()
corr = df[['Price (INR)', 'Resale Price (INR)']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

## Insight
- Prices are fairly spread across a wide range, suggesting diverse market segmentation from budget to premium bikes.

- There is a strong positive relationship between price and resale value, indicating that higher-priced bikes tend to retain higher resale values.

```python
# Registration Analysis
s_y = df.groupby("Registration Year").size().reset_index(name="Count")

s_y = s_y.sort_values("Registration Year") #sorting
sns.lineplot(data=s_y, x="Registration Year", y="Count", marker='o')

plt.xlabel("Year")
plt.ylabel("Number of Records")
plt.title("Bike Registrations Trend (Dataset)")
plt.show()
```

## Insight
- The dataset shows a strong upward trend in registrations over the years, especially after 2020, though this may reflect dataset distribution rather than actual market growth.
